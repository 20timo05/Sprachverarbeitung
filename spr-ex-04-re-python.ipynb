## Theory – Part 01

Give **at least one example and one counter-example** of full matches for each regular expression:

1. `r'TH(D|I|N|R)'`
2. `r'TH(D|I|N|R)+'`
3. `r'TH\s'`
4. `r'[1-9][0-9]+'`
5. `r'([^a-zA-Z])[tT]he([^a-zA-Z]) '`
6. `r'\b[0-9]+(\.[0-9]+)? *(km|[Kk]ilometers?)\b'`

````python
import re

# Fill in your examples
tests = {
    r"TH(D|I|N|R)": {"pos": ['THD'], "neg": ['THG']},
    r"TH(D|I|N|R)+": {"pos": ['THDI'], "neg": ['TH']},
    r"TH\s": {"pos": ['TH '], "neg": ['THD']},
    r"[1-9][0-9]+": {"pos": ['35'], "neg": ['0']},
    r"([^a-zA-Z]) [tT]he([^a-zA-Z]) ": {"pos": ['9 The8 '], "neg": ['aTheb']},
    r"\b[0-9]+(\.[0-9]+)? *(km|[Kk]ilometers?)\b": {"pos": ['8123.12345   Kilometer'], "neg": ['abc8123+bK']},
}

passed = total = 0

for pattern, data in tests.items():
    r = re.compile(pattern)
    print(f"\n{pattern}")

    for s in data["pos"]:
        total += 1
        ok = bool(r.fullmatch(s))
        print(" ✅" if ok else " ❌", s)
        passed += ok

    for s in data["neg"]:
        total += 1
        ok = not r.fullmatch(s)
        print(" ✅" if ok else " ❌", s)
        passed += ok

print(f"\nPassed {passed}/{total}")
````

### Part 02

Find regular expressions for the following types of strings:

* **Phone numbers** city codes including country prefix, e.g. `+42 001`
* **E-Mail addresses**, e.g. `test@test.de`
* **Mathematical expression of a fraction** that look like `23 / 42` (number operator number)
* **An case-sensitive identifier** in a programming language that looks like this: Start with a character from a to z or A to Z and continue with an arbitrary sequence of a character like the starting one or a decimal number. No other characters are allowed

````python
import re

# =========================
# ADD YOUR REGEX HERE
# =========================
patterns = {
    "phone": r"\+\d{1,4} \d+",      # TODO
    "email": r"\w+@\w+\.(de|com|org)",      # TODO
    "fraction": r"\d+ ?/ ?\d+",   # TODO
    "identifier": r"([a-zA-Z])(\1|0-9)*", # TODO
}

# ==========================================
# EXPANDED TEST DATA
# ==========================================
tests = {
    "phone": {
        "valid": [
            "+42 001",      # Standard match
            "+1 5551234",   # 1-digit country prefix
            "+4999 89"      # 4-digit country prefix
        ],
        "invalid": [
            "42 001",       # Missing '+'
            "+42001",       # Missing space
            "+42  001",     # Extra space (your pattern expects exactly 1 space)
            "+12345 001"    # Country prefix too long (> 4 digits)
        ]
    },
    "email": {
        "valid": [
            "test@test.de",
            "hello_world@domain.com",
            "user123@org.org"
        ],
        "invalid": [
            "test@test.net", # Unsupported domain suffix (only de|com|org allowed)
            "test@.de",      # Missing domain name
            "test.de",       # Missing '@'
            "@test.de",      # Missing username
            "test@test"      # Missing TLD
        ]
    },
    "fraction": {
        "valid": [
            "23 / 42",      # Spaces on both sides
            "23/42",        # No spaces
            "23   /   42"   # Multiple spaces
        ],
        "invalid": [
            "23/z42",       # Non-digit denominator
            "abc / 42",     # Non-digit numerator
            "23 / ",        # Missing denominator
            " / 42",        # Missing numerator
            "23.5 / 42"     # Contains decimals (only integers allowed)
        ]
    },
    "identifier": {
        "valid": [
            "var1",         # Normal alphanumeric
            "Test123",      # Normal alphanumeric starting with capital
            "a",            # 1-letter identifier (only works if patterns use *)
            "Z"             # 1-letter identifier (only works if patterns use *)
        ],
        "invalid": [
            "1var",         # Starts with a digit
            "var!",         # Contains invalid punctuation
            "test-1",       # Contains hyphen
            "",             # Empty string
            " "             # Space
        ]
    }
}

# ==========================================
# TEST RUNNER (SAME INTERFACE)
# ==========================================
for key in patterns:
    r = re.compile(patterns[key])
    print(f"\n==================== {key.upper()} ====================")

    print("  --- Valid Cases (Should Match) ---")
    for s in tests[key]["valid"]:
        print("  ✅" if r.fullmatch(s) else "  ❌", repr(s))

    print("  --- Invalid Cases (Should NOT Match) ---")
    for s in tests[key]["invalid"]:
        print("  ✅" if not r.fullmatch(s) else "  ❌", repr(s))
````

### Part 03

Create a list that contains at least the following words:

```python
[' wanted', ' learned ', ' justify ', ' justified ', 'read ', 'red ', 'get ', 'feed ', 'shed ', 'embed', 'proceed ']
```

Write a program that applies a regular expression on each word. The RE must remove a `-ed` suffix from a word. The output must be a new list with the changed words.

In order to make the program compact, use list comprehensions, see e.g., [W3Schools - Python List Comprehension](https://www.w3schools.com/python/python_lists_comprehension.asp).

````python
words = [' wanted', ' learned ', ' justify ', ' justified ', 'read ', 'red ', 'get ', 'feed ', 'shed ', 'embed', 'proceed ']


words_trimmed = [re.sub(r'(.*)(ed)( *$)', r'\1\3', w) for w in words]
words_trimmed
````

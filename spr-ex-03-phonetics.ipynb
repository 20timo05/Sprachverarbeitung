# Part 01: Analyze the production of sounds
Visit the website
https://en.wikipedia.org/wiki/International_Phonetic_Alphabet
Go to the section about consonants. Analyze the meaning of (at least) the following descriptions by comparing it with how you would produce the sound:  
• for place like  
&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;– Labial, Bilabial, Labiodental, Linguolabial  
&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;– Dental  
&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;– Palatal  
• for the manner  
&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;– Nasal  
&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;– Plosive  
&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;– Trill  
You may also learn about more terms about human sound production.

You may also learn about more terms about human sound production.

# Labial
- anything where the lips are used

# Bilabial
- anything where BOTH lips are used ('me', 'pull', 'box')

# Labiodental
- lower lip against upper teeth ('fine', 'vow')

# Linguolabial
- tongue against upper lip

# Dental
- tongue against front teeth

# Palatal
- tongue to the hard roof of the mouth ('yes')

# Nasal
- close mouth (either with lips or tongue), so that air has to go through nose ('nice', 'mice')

# Plosive
- block the air completely and then let it burst out at once ('p', 'b', 't')

# Trill
- let air pass through mouth and let tongue vibrate (rolled 'r')

# Part 01

Describe the difference between Syntax, Semantics and Pragmatics in linguistic studies. Explain an example for each.

1) Syntax: The rules that we have in a language for how the symbols must be structured & ORDERING, e.g. (where we have to place commas => WRONG), that we have to have a noun and a verb in a sentence, that sentences must end with a dot.

2) Semantics: It describes what the words actually mean. A sentence with a word that you don't know might sound right (because Syntax is intact), but you don't know the meaning of the word (because you don't know the semantics).

3) Pragmatics: The way that context influences meaning. Example: 
- Let's sit on the bank over there.
- I'm going to the bank to get some money.
=> The word bank has a different meaning within the two sentences.

- "It's cold in here!"
